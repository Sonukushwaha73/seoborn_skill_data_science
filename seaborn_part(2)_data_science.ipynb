# relplot to make a line plot

```python
import numpy as np
import  pandas as pd
time = np.arange(500)
value = np.random.randn(500).cumsum()# start se sab add karta hai
df= pd.DataFrame({"time_data":time, "value_data": value})
#time
df
```

```python
sns.relplot(x = "time_data",y = "value_data" ,data = df,kind = "line")
# | kind       | Plot        |
# | ---------- | ----------- |
# | `"bar"`    | Bar chart   |
# | `"count"`  | Count plot  |
# | `"box"`    | Box plot    |
# | `"violin"` | Violin plot |
# | `"strip"`  | Strip plot  |
# | `"swarm"`  | Swarm plot  |
# | `"point"`  | Point plot  |
```

```python
sns.relplot(x = "time_data",y = "value_data" ,data = df,kind = "line",sort = True)
```

```python
fmri = sns.load_dataset("fmri")
```

```python
fmri
```

```python
sns.relplot(x = "timepoint",  y = "subject",kind = "line",data = fmri )
plt.show()
```

```python
fmri.info()
```

```python
sns.relplot(x = "timepoint",  y = "subject",kind = "line",data = fmri,ci = "sd" )
# ci kya hota hai?

# ci ka full form hai Confidence Interval.

#  Ye batata hai data ke average (mean) ke around kitna uncertainty / variation hai.
#  Graph me ye shaded band / error bar ke form me dikhta hai.
plt.show()
```

```python
sns.relplot(x = "timepoint",  y = "subject",kind = "line",data = fmri,ci = "sd",estimator = None )
# estimator → value nikalega
plt.show()
```

```python
# another columns is add
sns.relplot(x = "timepoint",  y = "signal",kind = "line",data = fmri,ci = "sd",estimator = None,hue = "event",style = "region" )
# estimator → value nikalega
plt.show()
```

```python
sns.relplot(x = "timepoint",  y = "signal",kind = "line",data = fmri,ci = "sd",estimator = None,hue = "event",style = "region",markers = True,dashes = False )
# estimator → value nikalega
plt.show()
```
